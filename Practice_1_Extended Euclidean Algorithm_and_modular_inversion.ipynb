### Extended Euclidean Algorithm — Large Integer Optimized

```python
def extended_gcd(a, b):
    """
    Returns (gcd, x, y) such that a*x + b*y = gcd(a, b).
    Works efficiently even for very large integers.
    """
    old_r, r = a, b
    old_s, s = 1, 0
    old_t, t = 0, 1

    while r != 0:
        q = old_r // r
        old_r, r = r, old_r - q * r
        old_s, s = s, old_s - q * s
        old_t, t = t, old_t - q * t

    return old_r, old_s, old_t
```

```python
# Example: Compute gcd and Bézout coefficients with large integers
a = 2**2048 + 1
b = 2**1024 + 12345
gcd, x, y = extended_gcd(a, b)

# Now we Check if the computations are fine (they should satisfy ax+by=d:=gcd(a,b)):

print(f"gcd = {gcd}")
print(f"coefficients = {x,y}")
print(a*x+b*y)
print("\n Extended GCD has been successful for large integers.")
```

### Computation of inverses in modular arithmetic with large integers

```python
def modinv(a: int, m: int) -> int:
    """
    Modular inverse using the provided extended_gcd function.
    Takes a, m integers 
    Returns x such that (a * x) % m == 1.
    Raises ValueError if gcd(a, m) != 1.
    """
    g, x, _ = extended_gcd(a, m)
    if g != 1:
        raise ValueError("Inverse does not exist (a and m are not coprime).")
    return x % m

import secrets
#import math
# === Demonstration with large integers ===

BITS = 2048
while True:
    m = secrets.randbits(BITS) | 1      
    
# Generates a 2048-bit random number. The bitwise OR operation (| 1) ensures that the last bit is 1, 
# making $m$ always odd.
# Indeed: For the modular inverse to exist, we need gcd(a, m) = 1.
# If m were even, 50% of all possible values of a (all even numbers) 
# would share the common factor 2, automatically making gcd(a, m) \ge 2.
# By ensuring m is odd, the common factor 2 is ruled out, increasing the probability 
# that a random value of a is coprime to m in the while True loop (causing the loop to terminate faster).

# Now we generate a random integer with BITS - 1 bits:
    a = secrets.randbits(BITS - 1)

# And we bound 'a' to the range [2, m - 1], ensuring 1 < a < m:
    a = (a % (m - 2)) + 2

# Now we compute gcd(a,m): 
    g, _, _ = extended_gcd(a, m)
    if g == 1:             # need coprime
        break

inv = modinv(a, m)

print("Bit lengths: a =", a.bit_length(), "bits, m =", m.bit_length(), "bits")
print("Inverse verified:", (a * inv) % m == 1)

# Short previews so output isn’t huge
def short_hex(x):
    h = hex(x)[2:]
    return h if len(h) <= 32 else f"{h[:8]}…{h[-8:]}"

print("a   =", short_hex(a))
print("m   =", short_hex(m))
print("inv =", short_hex(inv))
```
