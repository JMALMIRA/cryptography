## Computing the Number of Irreducible Polynomials Over Finite Fields

This guide provides a clean, self-contained Python implementation for computing $N(m)$, the number of monic irreducible polynomials of degree $m$ over the finite field $\mathbb{Z}_p$ (or $GF(p)$), using the Möbius inversion formula we derived in the lectures (slides of week 4):

<strong>Mathematical Background</strong>

Recall from our algebraic derivation using the Tower Law (of finite extension fields) and root counting,  that the total number of elements in $GF(p^m)$ satisfies:

$$p^m = \sum_{k \mid m} k \cdot N(k)$$

Applying Möbius inversion to this identity yields the explicit formula for $N(m)$:

$$N(m) = \frac{1}{m} \sum_{d \mid m} \mu(d) p^{m/d}$$

where $\mu(d)$ is the classical Möbius function.

The following Python Code computes $N(m)$. First a helper function for the Möbius function is introduced. Then $N(m)$ is computed. 

```python
def mobius(n):
    """Computes the Möbius function mu(n)."""
    if n == 1:
        return 1
    
    # Find prime factorization to check for square-free and count distinct primes
    factors_count = 0
    temp = n
    d = 2
    while d * d <= temp:
        if temp % d == 0:
            factors_count += 1
            temp //= d
            if temp % d == 0:
                return 0  # Contains a squared prime factor (mu = 0)
        d += 1
    if temp > 1:
        factors_count += 1
        
    return (-1) ** factors_count


def count_irreducible_polynomials(p, m):
    """
    Computes N(m), the number of monic irreducible polynomials 
    of degree m over the finite field GF(p).
    
    Formula: N(m) = (1/m) * sum_{d | m} mu(d) * p^(m/d)
    """
    # Find all divisors of m
    divisors = [d for d in range(1, m + 1) if m % d == 0]
    
    # Apply the Möbius inversion formula sum
    total = sum(mobius(d) * (p ** (m // d)) for d in divisors)
    
    # N(m) must be an integer, so we use integer division
    return total // m
```

<strong> Example Usage & Verification</strong> 

You can run the following code to verify the values for $p = 2$ and $p = 101$ against your tables:

```python
# Test for p = 2
print("Number of irreducible polynomials for p = 2:")
for m in range(1, 11):
    print(f"  m = {m}: N({m}) = {count_irreducible_polynomials(2, m)}")

print("\n---------------------------------------------")

# Test for p = 101
print("Number of irreducible polynomials for p = 101:")
for m in range(1, 11):
    print(f"  m = {m}: N({m}) = {count_irreducible_polynomials(101, m)}")
```

### Let us compute a few primes...:

```python
def primes_smaller_than(n):
    """
    Returns a list of all prime numbers strictly smaller than n
    using the Sieve of Eratosthenes algorithm.
    """
    if n <= 2:
        return []
    
    # Initialize a boolean array "sieve" corresponding to numbers 0 to n-1
    sieve = [True] * n
    sieve[0] = sieve[1] = False  # 0 and 1 are not prime
    
    # Mark multiples of each prime as composite (False)
    for p in range(2, int(n ** 0.5) + 1):
        if sieve[p]:
            for i in range(p * p, n, p):
                sieve[i] = False
                
    # Collect all indices that remain True
    return [p for p in range(n) if sieve[p]]

primes_smaller_than(200)
```
